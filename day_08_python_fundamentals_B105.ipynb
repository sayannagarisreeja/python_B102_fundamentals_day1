```python
#continuation of dictionary datatype:
# general syntax of for loop:
fruits=['banana','apple','mango','cherry','guava']
mainvar=fruits
for tempvar in mainvar:
    print(tempvar)
```

```python
# key value pair .....> general syntax..>
items=['cake','chocolate','blackboard','duster']
mainvar=[items]
for items in mainvar:
   print(items)
```
